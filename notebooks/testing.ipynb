```python
from env.combat_env import AetherCombatEnv
from env.actions import EnemyAction

env = AetherCombatEnv()

obs, _ = env.reset()

print("Start:", env.enemy.x, env.enemy.z)

env.step(EnemyAction.MOVE_FORWARD)

print("Forward:", env.enemy.x, env.enemy.z)

env.reset()

env.step(EnemyAction.STRAFE_LEFT)

print("Left:", env.enemy.x, env.enemy.z)

env.reset()

env.step(EnemyAction.DODGE)

print("Dodge:", env.enemy.x, env.enemy.z)
```

```python
from env.actions import EnemyAction
from env.combat_env import AetherCombatEnv

env = AetherCombatEnv()

env.reset()

env.step(EnemyAction.FIREBALL)

print(len(env.projectiles))

for _ in range(5):
    env._update_projectiles()

for p in env.projectiles:
    print(p.x, p.z)
```

```python
from env.actions import EnemyAction
from env.combat_env import AetherCombatEnv

env = AetherCombatEnv()

env.reset()

print("Player HP:", env.player.hp)

for i in range(40):

    env.step(EnemyAction.FIREBALL)

    if i % 5 == 0:
        print(
            f"Step {i}",
            "HP:",
            env.player.hp,
            "Projectiles:",
            len(env.projectiles)
        )

print("Alive:", env.player.alive)
```
